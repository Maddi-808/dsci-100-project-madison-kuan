```r
ncaa_histo <- tidy_ncaa_data |>
    ggplot(aes(x = conf_wins)) +
        geom_histogram(bins = 10) +
        labs(x = "Number of Conference Wins", y = "Number of Players") +
        ggtitle(" Distribution of Number of Wins at the \n 2000 NCAA Conference") +
            theme(text = element_text(size = 20))
ncaa_histo
```

```r
#had to make a new tidy data set with two categorical values since the original only had one
#new tidy data has a cateogrical variable without levels and another categorical variable with levels
tidy_conf_data <- ncaa_data |>
    select(year,conf_wins, conf_losses, tourney_finish, conference) |> 
    filter(year %in% c("2000", "2001")) |> #added another year because in a competition there will likely be a set proportion 
                                            #of each category; adding another year will give more unknown information
    mutate(tourney_finish = as.factor(tourney_finish), conference = as.factor(conference)) 
#tidy_conf_data

options(repr.plot.width = 14, repr.plot.height = 8) #made sure proportions are wide so then you can really see each bar

ncaa_bar <- tidy_conf_data |>
    ggplot(aes(x = tourney_finish, fill = conference)) +
           geom_bar(position = "fill") +
        labs(x = "2000 NCAA Conference", fill = "Tourney Finish") +
        ggtitle("Proportion of Players from Different Conferences \nin the 2000 & 2001 Tourney Finish Categories") +
        theme(text = element_text(size = 20)) +
        theme(axis.text.x = element_text(angle = 20, hjust = 1)) 
ncaa_bar
```
